```python
import json
```

```python
json_data = [] # your list with json objects (dicts)

with open('track_id_and_artist_backup.json') as json_file:
   json_data = json.load(json_file)
```

```python
json_data
```

```python
clean_data = []
```

```python
for song in json_data:
    song_id = song[2]
    song_name = song[0]
    artists = []
    for artist in song[1]:
        artists.append(artist['name'])

    clean_data.append([song_id, song_name, artists])
```

```python
#clean_data
```

```python
with open("track_ids.json", "w") as f:
    json.dump(clean_data, f)
```
